```python

from itertools import product

def implies(a, b):
    return not a or b

print("P Q R | Q->P P->~Q Q|R | KB")

models = []

for P, Q, R in product([True, False], repeat=3):
    s1 = implies(Q, P)
    s2 = implies(P, not Q)
    s3 = Q or R
    KB = s1 and s2 and s3

    print(
        f"{'T' if P else 'F'} "
        f"{'T' if Q else 'F'} "
        f"{'T' if R else 'F'} | "
        f"  {'T' if s1 else 'F'}    "
        f"{'T' if s2 else 'F'}    "
        f"{'T' if s3 else 'F'} | "
        f"{'T' if KB else 'F'}"
    )

    if KB:
        models.append((P, Q, R))

def entails(query):
    return all(query(P, Q, R) for P, Q, R in models)

print("\nModels of KB:", models)

print("KB entails R:", entails(lambda P, Q, R: R))
print("KB entails R -> P:",
      entails(lambda P, Q, R: implies(R, P)))
print("KB entails Q -> R:",
      entails(lambda P, Q, R: implies(Q, R)))
```
