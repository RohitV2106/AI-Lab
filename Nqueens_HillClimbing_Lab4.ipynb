```python

def cost(state):
    n = len(state)
    conflicts = 0

    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def hill_climbing(state):
    n = len(state)

    while True:
        current_cost = cost(state)

        print("\nCurrent state:", state)
        print("Current cost:", current_cost)

        if current_cost == 0:
            print("Goal state reached!")
            break

        best_state = None
        best_cost = current_cost

        print("\nNeighbor states:")

        for i in range(n):
            for j in range(i + 1, n):
                neighbor = state.copy()
                neighbor[i], neighbor[j] = neighbor[j], neighbor[i]
                neighbor_cost = cost(neighbor)

                print("Swap", (i, j), ":", neighbor,
                      "Cost =", neighbor_cost)

                if neighbor_cost < best_cost:
                    best_cost = neighbor_cost
                    best_state = neighbor

        if best_state is None:
            print("Local minimum reached. No better neighbor.")
            break

        state = best_state


n = int(input("Enter number of queens: "))

if n < 1:
    print("Enter a positive number.")
else:
    print("Enter the initial row position of each queen")
    print("Use values from 0 to", n - 1)

    state = list(map(int, input().split()))

    if len(state) != n or any(x < 0 or x >= n for x in state):
        print("Invalid initial state!")
    else:
        hill_climbing(state)
```
